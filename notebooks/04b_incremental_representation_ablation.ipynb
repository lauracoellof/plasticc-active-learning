# 04b — Incremental E2-GP representation ablation

Use ten paired, stratified 80/20 splits of the same 5,820 training objects.
Hold the operational classifier fixed while adding GP1D, derived features and
GP2D in sequence. Report paired improvements with uncertainty and multiplicity
adjustment. The incremental contribution is conditional on this ordering.


```python
import sys
from pathlib import Path
ROOT = Path.cwd()
if not (ROOT / 'config.py').exists():
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from scripts.run_incremental_representation_ablation import run
RECOMPUTE = False
if RECOMPUTE:
    result_records, summary_table, contrastes = run()
else:
    from scripts.review_saved_ablations import review
    from scripts.run_incremental_representation_ablation import _paired_tests
    import pandas as pd
    import config
    result_records, summary_table = review('ablacion_causal_representacion', 'variante')
    contrastes = _paired_tests(result_records)
    saved = pd.read_csv(config.DIR_PROCESSED / 'ablacion_causal_representacion_contrastes.csv')
    numeric = contrastes.select_dtypes('number').columns
    pd.testing.assert_frame_equal(contrastes[numeric], saved[numeric], check_exact=False, rtol=1e-11, atol=1e-12)
    print('Paired contrasts also reproduced from saved per-seed scores.')
```
