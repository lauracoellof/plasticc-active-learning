# Appendix D — Photometric-representation sensitivity

Inspect how cleaning thresholds and GP1D temporal bounds affect retained
support and numerical fit diagnostics. These checks do not establish an
optimal predictive configuration.


```python
from pathlib import Path
import sys
import pandas as pd

ROOT = Path.cwd().resolve()
while not (ROOT / 'config.py').exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
import config
```

## Cleaning thresholds


```python
sens_limpieza = pd.read_csv(config.DIR_PROCESSED / 'sensibilidad_limpieza.csv')
display(sens_limpieza)
elegida = sens_limpieza.query('snr_min == 3 and min_obs == 5').iloc[0]
assert int(elegida.n_objetos) == 5820
assert int(elegida.bandas_validas_gp1d) == 12778
assert int(elegida.objetos_validos_gp2d) == 4699
```

### Support interpretation

Relaxing the object-level minimum can add objects without adding valid GP
fits, because GP1D and GP2D impose their own support requirements. A stricter
minimum removes both objects and potentially informative temporal coverage.


## Lower bound of the GP1D temporal length scale


```python
sens_ls = pd.read_csv(config.DIR_PROCESSED / 'sensibilidad_gp1d_ls.csv')
display(sens_ls)
assert sens_ls['n_ajustes'].eq(566).all()
assert sens_ls.loc[sens_ls.ls_min.eq(20), 'fraccion_en_cota'].iat[0] > 0.30
```

### Numerical interpretation

Compare marginal likelihood, residual summaries and the frequency of solutions
at the bound. Treat the ten-day choice as regularization in this experiment,
not as a physical time scale or universally optimal setting.
